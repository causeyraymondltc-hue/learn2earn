```python
import sqlite3

# Define the database file name
db_file = 'trainees.db'

try:
    # Connect to the SQLite database (it will be created if it doesn't exist)
    conn = sqlite3.connect(db_file)
    cursor = conn.cursor()

    # Create a table for trainees
    # We'll include columns like id, name, age, and course
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS trainees (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            age INTEGER,
            course TEXT
        )
    ''')
    print("Table 'trainees' checked/created successfully.")

    # --- Insert Sample Data --- #
    # Check if the table is empty before inserting to avoid duplicates on re-run
    cursor.execute('SELECT COUNT(*) FROM trainees')
    if cursor.fetchone()[0] == 0:
        trainees_data = [
            ('Alice Smith', 24, 'Python Programming'),
            ('Bob Johnson', 29, 'Data Science Fundamentals'),
            ('Charlie Brown', 22, 'Web Development Basics')
        ]
        cursor.executemany('INSERT INTO trainees (name, age, course) VALUES (?, ?, ?)', trainees_data)
        conn.commit()
        print("Sample trainee data inserted successfully.")
    else:
        print("Trainee data already exists, skipping sample insertion.")

    # --- Retrieve and Display Data --- #
    print("\n--- Current Trainees ---")
    cursor.execute('SELECT id, name, age, course FROM trainees')
    rows = cursor.fetchall()

    if rows:
        for row in rows:
            print(f"ID: {row[0]}, Name: {row[1]}, Age: {row[2]}, Course: {row[3]}")
    else:
        print("No trainees found in the database.")

except sqlite3.Error as e:
    print(f"An SQLite error occurred: {e}")
finally:
    # Close the connection
    if conn:
        conn.close()
        print(f"\nDatabase '{db_file}' connection closed.")

# To add more trainees later, you would use:
# new_trainee = ('Diana Prince', 27, 'Machine Learning')
# cursor.execute('INSERT INTO trainees (name, age, course) VALUES (?, ?, ?)', new_trainee)
# conn.commit()

# To query specific trainees:
# cursor.execute('SELECT * FROM trainees WHERE course = ?', ('Python Programming',))
# print(cursor.fetchall())
```
